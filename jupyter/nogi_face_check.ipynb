```python
import face_recognition
import cv2
import os
import matplotlib.pyplot as plt

# 既知の顔データを読み込む
known_face_encodings = []
known_face_names = []

import os
import numpy as np
import face_recognition
import cv2
import matplotlib.pyplot as plt

known_face_encodings = []
known_face_names = []

def load_known_faces(folder_path: str):
    """フォルダ内の既知顔を読み込み。RGB/uint8に正規化。"""
    known_face_encodings.clear()
    known_face_names.clear()
    for fname in os.listdir(folder_path):
        if not fname.lower().endswith((".jpg", ".jpeg", ".png")):
            continue
        path = os.path.join(folder_path, fname)
        img = face_recognition.load_image_file(path)      # ← RGB, dtype=uint8 のはず
        img = np.asarray(img, dtype=np.uint8)             # 念のため dtype を固定
        if img.ndim == 2:                                 # グレースケールなら RGB 化
            img = np.stack([img]*3, axis=-1)
        elif img.shape[-1] == 4:                          # RGBA → RGB
            img = img[:, :, :3]

        encs = face_recognition.face_encodings(img)
        if len(encs) > 0:
            known_face_encodings.append(encs[0])
            known_face_names.append(os.path.splitext(fname)[0])
    print(f"Loaded {len(known_face_names)} known faces: {known_face_names}")

def recognize_faces_in_image(image_path: str, tolerance: float = 0.6, show=True):
    """単一画像の照合。類似度も計算して出力。"""
    img = face_recognition.load_image_file(image_path)
    img = np.asarray(img, dtype=np.uint8)
    if img.ndim == 2:
        img = np.stack([img]*3, axis=-1)
    elif img.shape[-1] == 4:
        img = img[:, :, :3]

    # 顔検出 & エンコード
    face_locations = face_recognition.face_locations(img)
    face_encodings  = face_recognition.face_encodings(img, face_locations)

    results = []
    for encoding, (top, right, bottom, left) in zip(face_encodings, face_locations):
        name = "Unknown"
        similarity = 0.0

        if known_face_encodings:
            dists = face_recognition.face_distance(known_face_encodings, encoding)
            best_idx = int(np.argmin(dists))
            best_dist = dists[best_idx]
            matches = face_recognition.compare_faces(known_face_encodings, encoding, tolerance=tolerance)
            if matches[best_idx]:
                name = known_face_names[best_idx]
            similarity = max(0.0, 1 - best_dist)  # 1に近いほど似ている
        results.append((name, similarity, (top, right, bottom, left)))

    # 結果を描画（matplotlib）
    if show:
        img_draw = img.copy()
        for name, sim, (top, right, bottom, left) in results:
            label = f"{name} ({sim*100:.1f}%)"
            cv2.rectangle(img_draw, (left, top), (right, bottom), (0, 255, 0), 2)
            cv2.putText(img_draw, label, (left, max(0, top-10)),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)
        plt.figure(figsize=(8, 6))
        plt.imshow(img_draw)
        plt.axis("off")
        plt.title("Recognition Result with Similarity")
        plt.show()

    return results


```

```python
from pathlib import Path
import numpy as np
import cv2
import face_recognition
import matplotlib.pyplot as plt

# ===== 共通ユーティリティ =====
def ensure_rgb_uint8(img):
    arr = np.asarray(img)
    if arr.dtype != np.uint8:
        arr = arr.astype(np.uint8)
    if arr.ndim == 2:
        arr = np.stack([arr]*3, axis=-1)
    elif arr.shape[-1] == 4:
        arr = arr[:, :, :3]
    return arr

def draw_boxes_rgb(img_rgb, results, title=None):
    canvas = img_rgb.copy()
    for label, score, (top, right, bottom, left) in results:
        cv2.rectangle(canvas, (left, top), (right, bottom), (0, 255, 0), 2)
        txt = f"{label} ({score*100:.1f}%)"
        cv2.putText(canvas, txt, (left, max(0, top-8)),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0), 2)
    plt.figure(figsize=(8,6))
    plt.imshow(canvas)
    plt.axis("off")
    if title: plt.title(title)
    plt.show()

def l2_normalize(v, eps=1e-12):
    n = np.linalg.norm(v, axis=-1, keepdims=True)
    return v / np.clip(n, eps, None)

def cosine_sim(a, b):
    a = l2_normalize(a)
    b = l2_normalize(b)
    return np.dot(a, b.T)

# ===== ここを強化：known_faces の収集 =====
def collect_known_images(known_dir: Path):
    """
    known_faces/ の中から人物→画像パスの辞書を作る。
    - サブフォルダ（推奨）: known_faces/alice/*.jpg, *.png ...
    - 直下ファイル: known_faces/alice.jpg なども許可
    - 隠しフォルダ/ファイル（.で始まる）は除外
    """
    if not known_dir.exists():
        raise FileNotFoundError(f"{known_dir} が存在しません")

    people = {}
    valid_exts = {".jpg", ".jpeg", ".png"}

    # サブフォルダ優先
    subdirs = [p for p in known_dir.iterdir() if p.is_dir() and not p.name.startswith(".")]
    if subdirs:
        for person_dir in sorted(subdirs):
            imgs = []
            for ext in valid_exts:
                imgs += list(person_dir.glob(f"*{ext}"))
            # 直下にドット始まりのファイルは拾わない
            imgs = [p for p in imgs if not p.name.startswith(".")]
            if imgs:
                people[person_dir.name] = imgs
    else:
        # 直下の画像を「ファイル名=人名」として扱う
        imgs = [p for p in known_dir.iterdir()
                if p.is_file() and p.suffix.lower() in valid_exts and not p.name.startswith(".")]
        for p in imgs:
            people.setdefault(p.stem, []).append(p)

    # デバッグ出力
    total_imgs = sum(len(v) for v in people.values())
    print(f"[collect_known_images] persons={len(people)}, images={total_imgs}")
    for k, v in people.items():
        print(f"  - {k}: {len(v)} files")
    return people

# ===== dlib(face_recognition) =====
def build_gallery_face_recognition(known_dir: Path):
    people = collect_known_images(known_dir)
    templates = {}
    for name, paths in people.items():
        vecs = []
        for path in paths:
            img = face_recognition.load_image_file(str(path))
            img = ensure_rgb_uint8(img)
            encs = face_recognition.face_encodings(img)
            if len(encs) > 0:
                vecs.append(encs[0])
        if vecs:
            v = np.mean(np.vstack(vecs), axis=0)
            templates[name] = l2_normalize(v)
    print(f"[dlib] templates built: {list(templates.keys())}")
    return templates

def recognize_with_face_recognition(image_path: Path, templates: dict):
    img = face_recognition.load_image_file(str(image_path))
    img = ensure_rgb_uint8(img)
    face_locations = face_recognition.face_locations(img)
    face_encs = face_recognition.face_encodings(img, face_locations)

    names = list(templates.keys())
    gal = np.vstack([templates[n] for n in names]) if names else None

    results = []
    for enc, (top, right, bottom, left) in zip(face_encs, face_locations):
        label = "Unknown"; sim_best = 0.0
        if gal is not None and len(gal) > 0:
            sims = cosine_sim(l2_normalize(enc[None, :]), gal)[0]
            best_idx = int(np.argmax(sims))
            sim_best = float(sims[best_idx])
            label = names[best_idx]
        results.append((label, sim_best, (top, right, bottom, left)))
    return img, results

# ===== InsightFace =====
from insightface.app import FaceAnalysis

class InsightFaceWrapper:
    def __init__(self, ctx_id=0, det_size=(640,640), model_pack="buffalo_l"):
        self.app = FaceAnalysis(name=model_pack)
        self.app.prepare(ctx_id=ctx_id, det_size=det_size)

    def embed_bgr(self, bgr):
        faces = self.app.get(bgr)
        embs, boxes = [], []
        for f in faces:
            embs.append(f.normed_embedding)  # L2済み
            boxes.append(f.bbox.astype(int))
        if not embs:
            return np.zeros((0,512), dtype=np.float32), []
        return np.stack(embs), boxes

def build_gallery_insightface(known_dir: Path, ins: InsightFaceWrapper):
    people = collect_known_images(known_dir)
    templates = {}
    for name, paths in people.items():
        vecs = []
        for path in paths:
            bgr = cv2.imread(str(path))
            if bgr is None:
                continue
            embs, _ = ins.embed_bgr(bgr)
            if len(embs):
                vecs.append(embs[0])
        if vecs:
            v = np.mean(np.vstack(vecs), axis=0)
            templates[name] = l2_normalize(v)
    print(f"[insightface] templates built: {list(templates.keys())}")
    return templates

def recognize_with_insightface(image_path: Path, templates: dict, ins: InsightFaceWrapper):
    bgr = cv2.imread(str(image_path))
    if bgr is None:
        raise FileNotFoundError(image_path)
    embs, boxes = ins.embed_bgr(bgr)
    names = list(templates.keys())
    gal = np.vstack([templates[n] for n in names]) if names else None

    results = []
    for e, (l, t, r, b) in zip(embs, boxes):
        label = "Unknown"; sim_best = 0.0
        if gal is not None and len(gal) > 0:
            sims = np.dot(l2_normalize(e[None,:]), gal.T)[0]
            best_idx = int(np.argmax(sims))
            sim_best = float(sims[best_idx])
            label = names[best_idx]
        results.append((label, sim_best, (int(t), int(r), int(b), int(l))))  # to (top,right,bottom,left)
    img_rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
    return img_rgb, results

# ===== 比較ランナー =====

def compare_both(gal_dlib, gal_ins, image_path="test_image.jpg",
                 dlib_sim_thr=0.65, insight_sim_thr=0.70):
    # known_dir = Path(known_dir)
    image_path = Path(image_path)

    # dlib
    gal_dlib = build_gallery_face_recognition(known_dir)
    imgA, resA = recognize_with_face_recognition(image_path, gal_dlib)
    resA_thr = [(name if sim>=dlib_sim_thr else "Unknown", sim, bbox) for (name, sim, bbox) in resA]

    # InsightFace
    imgB, resB = recognize_with_insightface(image_path, gal_ins, ins)
    resB_thr = [(name if sim>=insight_sim_thr else "Unknown", sim, bbox) for (name, sim, bbox) in resB]

    print("=== dlib(face_recognition) results ===")
    for name, sim, _ in resA_thr:
        print(f"  {name:>16s} : {sim*100:6.2f}%")
    print("=== InsightFace(ArcFace) results ===")
    for name, sim, _ in resB_thr:
        print(f"  {name:>16s} : {sim*100:6.2f}%")

    draw_boxes_rgb(imgA, resA_thr, title="dlib (face_recognition) - cosine similarity")
    draw_boxes_rgb(imgB, resB_thr, title="InsightFace (RetinaFace + ArcFace) - cosine similarity")
```

```python
# =======================
# All-in-One cell for Jupyter
# =======================
from pathlib import Path
import os
import numpy as np
import cv2
import matplotlib.pyplot as plt
import face_recognition

# ========== 基本ユーティリティ ==========
def ensure_rgb_uint8(img):
    """入力を (H,W,3) の RGB uint8 に整形"""
    arr = np.asarray(img)
    if arr.ndim == 2:
        arr = np.stack([arr]*3, axis=-1)
    elif arr.ndim == 3 and arr.shape[2] == 4:
        arr = arr[:, :, :3]  # RGBA -> RGB
    if arr.dtype != np.uint8:
        arr = arr.astype(np.uint8, copy=False)
    if not arr.flags["C_CONTIGUOUS"]:
        arr = np.ascontiguousarray(arr)
    return arr

def draw_boxes_rgb(img_rgb, results, title=None):
    """results: [(label, score, (top,right,bottom,left))...] をRGB画像に描画して表示"""
    canvas = img_rgb.copy()
    for label, score, (top, right, bottom, left) in results:
        cv2.rectangle(canvas, (left, top), (right, bottom), (0, 255, 0), 2)
        txt = f"{label} ({score*100:.1f}%)"
        cv2.putText(canvas, txt, (left, max(0, top-8)),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0), 2)
    plt.figure(figsize=(8,6))
    plt.imshow(canvas)
    plt.axis("off")
    if title: plt.title(title)
    plt.show()

def l2_normalize(v, eps=1e-12):
    v = np.asarray(v)
    n = np.linalg.norm(v, axis=-1, keepdims=True)
    return v / np.clip(n, eps, None)

def cosine_sim(a, b):
    """a: (..., D), b: (N, D) -> dot(a, b.T)（どちらもL2前提で内積=cos）"""
    a = l2_normalize(a)
    b = l2_normalize(b)
    return np.dot(a, b.T)

# ========== known_faces の収集（隠しフォルダ/ファイル除外） ==========
def collect_known_images(known_dir: Path):
    """
    known_faces/ の中から人物→画像パスの辞書を作る。
    - サブフォルダ推奨: known_faces/alice/*.jpg, *.png ...
    - 直下ファイル: known_faces/alice.jpg なども許可
    - 隠しフォルダ/ファイル（.で始まる）は除外
    """
    if not known_dir.exists():
        raise FileNotFoundError(f"{known_dir} が存在しません")

    people = {}
    valid_exts = {".jpg", ".jpeg", ".png"}

    subdirs = [p for p in known_dir.iterdir() if p.is_dir() and not p.name.startswith(".")]
    if subdirs:
        for person_dir in sorted(subdirs):
            imgs = []
            for ext in valid_exts:
                imgs += list(person_dir.glob(f"*{ext}"))
            imgs = [p for p in imgs if not p.name.startswith(".")]
            if imgs:
                people[person_dir.name] = imgs
    else:
        imgs = [p for p in known_dir.iterdir()
                if p.is_file() and p.suffix.lower() in valid_exts and not p.name.startswith(".")]
        for p in imgs:
            people.setdefault(p.stem, []).append(p)

    # デバッグ出力
    total_imgs = sum(len(v) for v in people.values())
    print(f"[collect_known_images] persons={len(people)}, images={total_imgs}")
    for k, v in people.items():
        print(f"  - {k}: {len(v)} files")
    return people

# ==========================================================
#  A) dlib / face_recognition パイプライン（安全版）
#     - C-contiguous / RGB uint8
#     - flip TTA
#     - コサイン類似度
# ==========================================================
def safe_face_locations(img_rgb):
    """face_locations を int タプル化して返す"""
    locs = face_recognition.face_locations(img_rgb)
    return [(int(t), int(r), int(b), int(l)) for (t,r,b,l) in locs]

def encode_dlib_rgb_safe(img_rgb, face_locations=None, num_jitters=3, do_flip_tta=True):
    """dlib埋め込み（RGB/uint8/C-contiguous前提） + 水平反転TTA（copy()で連続確保）"""
    img_rgb = ensure_rgb_uint8(img_rgb)

    if face_locations is None:
        face_locations = safe_face_locations(img_rgb)
    else:
        face_locations = [(int(t), int(r), int(b), int(l)) for (t,r,b,l) in face_locations]

    encs = face_recognition.face_encodings(img_rgb, face_locations, num_jitters=num_jitters)
    encs = np.asarray(encs, dtype=np.float32)

    if do_flip_tta and len(encs):
        img_flip = img_rgb[:, ::-1, :].copy()  # C連続を確保
        w = img_rgb.shape[1]
        locs_flip = [(t, w - l, b, w - r) for (t,r,b,l) in face_locations]
        locs_flip = [(int(t), int(r), int(b), int(l)) for (t,r,b,l) in locs_flip]
        encs_flip = face_recognition.face_encodings(img_flip, locs_flip, num_jitters=1)
        if len(encs_flip) == len(encs):
            encs = (encs + np.asarray(encs_flip, dtype=np.float32)) / 2.0

    encs = l2_normalize(encs)
    return encs, face_locations

def build_gallery_face_recognition(known_dir: Path, num_jitters=1):
    people = collect_known_images(known_dir)
    templates = {}  # name -> 128D (L2済み)
    for name, paths in people.items():
        vecs = []
        for path in paths:
            img = face_recognition.load_image_file(str(path))
            img = ensure_rgb_uint8(img)
            locs = safe_face_locations(img)
            if not locs: 
                continue
            encs, _ = encode_dlib_rgb_safe(img, face_locations=locs, num_jitters=num_jitters, do_flip_tta=True)
            if len(encs):  # 同一画像内に複数顔があれば平均
                vecs.append(np.mean(encs, axis=0))
        if vecs:
            v = np.mean(np.vstack(vecs), axis=0)
            templates[name] = l2_normalize(v)
    print(f"[dlib] templates built: {list(templates.keys())}")
    return templates

def recognize_with_face_recognition(image_path: Path, templates: dict, num_jitters=3):
    img = face_recognition.load_image_file(str(image_path))
    img = ensure_rgb_uint8(img)
    encs, face_locations = encode_dlib_rgb_safe(img, num_jitters=num_jitters, do_flip_tta=True)

    names = list(templates.keys())
    gal = np.vstack([templates[n] for n in names]) if names else None

    results = []
    for q, (top, right, bottom, left) in zip(encs, face_locations):
        label = "Unknown"; sim_best = 0.0
        if gal is not None and len(gal) > 0:
            sims = cosine_sim(q[None, :], gal)[0]
            j = int(np.argmax(sims))
            sim_best = float(sims[j])
            label = names[j]
        results.append((label, sim_best, (top, right, bottom, left)))
    return img, results

# ==========================================================
#  B) InsightFace (RetinaFace + ArcFace) パイプライン
#     - FaceAnalysis
#     - テンプレ平均
#     - flip TTA（簡易）
# ==========================================================
try:
    from insightface.app import FaceAnalysis
except Exception as e:
    raise RuntimeError(
        "InsightFaceのインポートに失敗。まず `pip install insightface onnxruntime-gpu` "
        "（GPUなしなら onnxruntime）を実行してください。\n原文:\n" + repr(e)
    )

class InsightFaceWrapper:
    def __init__(self, ctx_id=0, det_size=(640, 640), model_pack="buffalo_l"):
        # GPU: ctx_id=0 / CPU: ctx_id=-1
        self.app = FaceAnalysis(name=model_pack)
        self.app.prepare(ctx_id=ctx_id, det_size=det_size)

    def embed_bgr(self, bgr, do_flip_tta=True):
        faces = self.app.get(bgr)
        embs, boxes = [], []
        for f in faces:
            e = f.normed_embedding.astype(np.float32)  # L2済み
            if do_flip_tta:
                bgrf = cv2.flip(bgr, 1).copy()
                faces_f = self.app.get(bgrf)
                if faces_f:
                    e = (e + faces_f[0].normed_embedding.astype(np.float32)) / 2.0
                    e = e / (np.linalg.norm(e) + 1e-12)
            embs.append(e)
            boxes.append(f.bbox.astype(int))  # (l,t,r,b)
        if not embs:
            return np.zeros((0,512), np.float32), []
        return np.stack(embs), boxes

def build_gallery_insightface(known_dir: Path, ins: InsightFaceWrapper):
    people = collect_known_images(known_dir)
    templates = {}  # name -> 512D (L2済み)
    for name, paths in people.items():
        vecs = []
        for path in paths:
            bgr = cv2.imread(str(path))
            if bgr is None:
                continue
            embs, _ = ins.embed_bgr(bgr, do_flip_tta=True)
            if len(embs):
                vecs.append(embs[0])
        if vecs:
            v = np.mean(np.vstack(vecs), axis=0)
            templates[name] = l2_normalize(v)
    print(f"[insightface] templates built: {list(templates.keys())}")
    return templates

def recognize_with_insightface(image_path: Path, templates: dict, ins: InsightFaceWrapper):
    bgr = cv2.imread(str(image_path))
    if bgr is None:
        raise FileNotFoundError(image_path)
    embs, boxes = ins.embed_bgr(bgr, do_flip_tta=True)

    names = list(templates.keys())
    gal = np.vstack([templates[n] for n in names]) if names else None

    results = []
    for e, (l, t, r, b) in zip(embs, boxes):
        label = "Unknown"; sim_best = 0.0
        if gal is not None and len(gal) > 0:
            sims = np.dot(l2_normalize(e[None,:]), gal.T)[0]  # cos sim
            j = int(np.argmax(sims))
            sim_best = float(sims[j])
            label = names[j]
        results.append((label, sim_best, (int(t), int(r), int(b), int(l))))  # (top,right,bottom,left)
    img_rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
    return img_rgb, results

# ==========================================================
#  比較ランナー（閾値適用して Unknown 落とし）
# ==========================================================
def compare_both(known_dir="known_faces", image_path="test_image.jpg",
                 dlib_sim_thr=0.70, insight_sim_thr=0.70,
                 insight_ctx_id=0, insight_det_size=(640,640)):
    known_dir = Path(known_dir)
    image_path = Path(image_path)

    # dlib
    gal_dlib = build_gallery_face_recognition(known_dir, num_jitters=1)
    imgA, resA = recognize_with_face_recognition(image_path, gal_dlib, num_jitters=3)
    resA_thr = [(name if sim>=dlib_sim_thr else "Unknown", sim, bbox) for (name, sim, bbox) in resA]

    # InsightFace
    ins = InsightFaceWrapper(ctx_id=insight_ctx_id, det_size=insight_det_size)
    gal_ins = build_gallery_insightface(known_dir, ins)
    imgB, resB = recognize_with_insightface(image_path, gal_ins, ins)
    resB_thr = [(name if sim>=insight_sim_thr else "Unknown", sim, bbox) for (name, sim, bbox) in resB]

    print("=== dlib(face_recognition) results ===")
    for name, sim, _ in resA_thr:
        print(f"  {name:>16s} : {sim*100:6.2f}%")
    print("=== InsightFace(ArcFace) results ===")
    for name, sim, _ in resB_thr:
        print(f"  {name:>16s} : {sim*100:6.2f}%")

    draw_boxes_rgb(imgA, resA_thr, title="dlib (face_recognition) - cosine similarity")
    draw_boxes_rgb(imgB, resB_thr, title="InsightFace (RetinaFace + ArcFace) - cosine similarity")

```

```python
# =======================
# All-in-One cell for Jupyter
# =======================
from pathlib import Path
import os
import numpy as np
import cv2
import matplotlib.pyplot as plt
import face_recognition

# ========== 基本ユーティリティ ==========
def ensure_rgb_uint8(img):
    """入力を (H,W,3) の RGB uint8 に整形"""
    arr = np.asarray(img)
    if arr.ndim == 2:
        arr = np.stack([arr]*3, axis=-1)
    elif arr.ndim == 3 and arr.shape[2] == 4:
        arr = arr[:, :, :3]  # RGBA -> RGB
    if arr.dtype != np.uint8:
        arr = arr.astype(np.uint8, copy=False)
    if not arr.flags["C_CONTIGUOUS"]:
        arr = np.ascontiguousarray(arr)
    return arr

def draw_boxes_rgb(img_rgb, results, title=None):
    """results: [(label, score, (top,right,bottom,left))...] をRGB画像に描画して表示"""
    canvas = img_rgb.copy()
    for label, score, (top, right, bottom, left) in results:
        cv2.rectangle(canvas, (left, top), (right, bottom), (0, 255, 0), 2)
        txt = f"{label} ({score*100:.1f}%)"
        cv2.putText(canvas, txt, (left, max(0, top-8)),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0), 2)
    plt.figure(figsize=(8,6))
    plt.imshow(canvas)
    plt.axis("off")
    if title: plt.title(title)
    plt.show()

def l2_normalize(v, eps=1e-12):
    v = np.asarray(v)
    n = np.linalg.norm(v, axis=-1, keepdims=True)
    return v / np.clip(n, eps, None)

def cosine_sim(a, b):
    """a: (..., D), b: (N, D) -> dot(a, b.T)（どちらもL2前提で内積=cos）"""
    a = l2_normalize(a)
    b = l2_normalize(b)
    return np.dot(a, b.T)

# ========== known_faces の収集（隠しフォルダ/ファイル除外） ==========
def collect_known_images(known_dir: Path):
    """
    known_faces/ の中から人物→画像パスの辞書を作る。
    - サブフォルダ推奨: known_faces/alice/*.jpg, *.png ...
    - 直下ファイル: known_faces/alice.jpg なども許可
    - 隠しフォルダ/ファイル（.で始まる）は除外
    """
    if not known_dir.exists():
        raise FileNotFoundError(f"{known_dir} が存在しません")

    people = {}
    valid_exts = {".jpg", ".jpeg", ".png"}

    subdirs = [p for p in known_dir.iterdir() if p.is_dir() and not p.name.startswith(".")]
    if subdirs:
        for person_dir in sorted(subdirs):
            imgs = []
            for ext in valid_exts:
                imgs += list(person_dir.glob(f"*{ext}"))
            imgs = [p for p in imgs if not p.name.startswith(".")]
            if imgs:
                people[person_dir.name] = imgs
    else:
        imgs = [p for p in known_dir.iterdir()
                if p.is_file() and p.suffix.lower() in valid_exts and not p.name.startswith(".")]
        for p in imgs:
            people.setdefault(p.stem, []).append(p)

    # デバッグ出力
    total_imgs = sum(len(v) for v in people.values())
    print(f"[collect_known_images] persons={len(people)}, images={total_imgs}")
    for k, v in people.items():
        print(f"  - {k}: {len(v)} files")
    return people

# ==========================================================
#  A) dlib / face_recognition パイプライン（安全版）
#     - C-contiguous / RGB uint8
#     - flip TTA
#     - コサイン類似度
# ==========================================================
def safe_face_locations(img_rgb):
    """face_locations を int タプル化して返す"""
    locs = face_recognition.face_locations(img_rgb)
    return [(int(t), int(r), int(b), int(l)) for (t,r,b,l) in locs]

def encode_dlib_rgb_safe(img_rgb, face_locations=None, num_jitters=3, do_flip_tta=True):
    """dlib埋め込み（RGB/uint8/C-contiguous前提） + 水平反転TTA（copy()で連続確保）"""
    img_rgb = ensure_rgb_uint8(img_rgb)

    if face_locations is None:
        face_locations = safe_face_locations(img_rgb)
    else:
        face_locations = [(int(t), int(r), int(b), int(l)) for (t,r,b,l) in face_locations]

    encs = face_recognition.face_encodings(img_rgb, face_locations, num_jitters=num_jitters)
    encs = np.asarray(encs, dtype=np.float32)

    if do_flip_tta and len(encs):
        img_flip = img_rgb[:, ::-1, :].copy()  # C連続を確保
        w = img_rgb.shape[1]
        locs_flip = [(t, w - l, b, w - r) for (t,r,b,l) in face_locations]
        locs_flip = [(int(t), int(r), int(b), int(l)) for (t,r,b,l) in locs_flip]
        encs_flip = face_recognition.face_encodings(img_flip, locs_flip, num_jitters=1)
        if len(encs_flip) == len(encs):
            encs = (encs + np.asarray(encs_flip, dtype=np.float32)) / 2.0

    encs = l2_normalize(encs)
    return encs, face_locations

def build_gallery_face_recognition(known_dir: Path, num_jitters=1):
    people = collect_known_images(known_dir)
    templates = {}  # name -> 128D (L2済み)
    for name, paths in people.items():
        vecs = []
        for path in paths:
            img = face_recognition.load_image_file(str(path))
            img = ensure_rgb_uint8(img)
            locs = safe_face_locations(img)
            if not locs: 
                continue
            encs, _ = encode_dlib_rgb_safe(img, face_locations=locs, num_jitters=num_jitters, do_flip_tta=True)
            if len(encs):  # 同一画像内に複数顔があれば平均
                vecs.append(np.mean(encs, axis=0))
        if vecs:
            v = np.mean(np.vstack(vecs), axis=0)
            templates[name] = l2_normalize(v)
    print(f"[dlib] templates built: {list(templates.keys())}")
    return templates

def recognize_with_face_recognition(image_path: Path, templates: dict, num_jitters=3):
    img = face_recognition.load_image_file(str(image_path))
    img = ensure_rgb_uint8(img)
    encs, face_locations = encode_dlib_rgb_safe(img, num_jitters=num_jitters, do_flip_tta=True)

    names = list(templates.keys())
    gal = np.vstack([templates[n] for n in names]) if names else None

    results = []
    for q, (top, right, bottom, left) in zip(encs, face_locations):
        label = "Unknown"; sim_best = 0.0
        if gal is not None and len(gal) > 0:
            sims = cosine_sim(q[None, :], gal)[0]
            j = int(np.argmax(sims))
            sim_best = float(sims[j])
            label = names[j]
        results.append((label, sim_best, (top, right, bottom, left)))
    return img, results

# ==========================================================
#  B) InsightFace (RetinaFace + ArcFace) パイプライン
#     - FaceAnalysis
#     - テンプレ平均
#     - flip TTA（簡易）
# ==========================================================
try:
    from insightface.app import FaceAnalysis
except Exception as e:
    raise RuntimeError(
        "InsightFaceのインポートに失敗。まず `pip install insightface onnxruntime-gpu` "
        "（GPUなしなら onnxruntime）を実行してください。\n原文:\n" + repr(e)
    )

class InsightFaceWrapper:
    def __init__(self, ctx_id=0, det_size=(640, 640), model_pack="buffalo_l"):
        # GPU: ctx_id=0 / CPU: ctx_id=-1
        self.app = FaceAnalysis(name=model_pack)
        self.app.prepare(ctx_id=ctx_id, det_size=det_size)

    def embed_bgr(self, bgr, do_flip_tta=True):
        faces = self.app.get(bgr)
        embs, boxes = [], []
        for f in faces:
            e = f.normed_embedding.astype(np.float32)  # L2済み
            if do_flip_tta:
                bgrf = cv2.flip(bgr, 1).copy()
                faces_f = self.app.get(bgrf)
                if faces_f:
                    e = (e + faces_f[0].normed_embedding.astype(np.float32)) / 2.0
                    e = e / (np.linalg.norm(e) + 1e-12)
            embs.append(e)
            boxes.append(f.bbox.astype(int))  # (l,t,r,b)
        if not embs:
            return np.zeros((0,512), np.float32), []
        return np.stack(embs), boxes

def build_gallery_insightface(known_dir: Path, ins: InsightFaceWrapper):
    people = collect_known_images(known_dir)
    templates = {}  # name -> 512D (L2済み)
    for name, paths in people.items():
        vecs = []
        for path in paths:
            bgr = cv2.imread(str(path))
            if bgr is None:
                continue
            embs, _ = ins.embed_bgr(bgr, do_flip_tta=True)
            if len(embs):
                vecs.append(embs[0])
        if vecs:
            v = np.mean(np.vstack(vecs), axis=0)
            templates[name] = l2_normalize(v)
    print(f"[insightface] templates built: {list(templates.keys())}")
    return templates

def recognize_with_insightface(image_path: Path, templates: dict, ins: InsightFaceWrapper):
    bgr = cv2.imread(str(image_path))
    if bgr is None:
        raise FileNotFoundError(image_path)
    embs, boxes = ins.embed_bgr(bgr, do_flip_tta=True)

    names = list(templates.keys())
    gal = np.vstack([templates[n] for n in names]) if names else None

    results = []
    for e, (l, t, r, b) in zip(embs, boxes):
        label = "Unknown"; sim_best = 0.0
        if gal is not None and len(gal) > 0:
            sims = np.dot(l2_normalize(e[None,:]), gal.T)[0]  # cos sim
            j = int(np.argmax(sims))
            sim_best = float(sims[j])
            label = names[j]
        results.append((label, sim_best, (int(t), int(r), int(b), int(l))))  # (top,right,bottom,left)
    img_rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
    return img_rgb, results

# ==========================================================
#  比較ランナー（閾値適用して Unknown 落とし）
# ==========================================================
def compare_both(known_dir="known_faces", image_path="test_image.jpg",
                 dlib_sim_thr=0.70, insight_sim_thr=0.70,
                 insight_ctx_id=0, insight_det_size=(640,640)):
    known_dir = Path(known_dir)
    image_path = Path(image_path)

    # dlib
    gal_dlib = build_gallery_face_recognition(known_dir, num_jitters=1)
    imgA, resA = recognize_with_face_recognition(image_path, gal_dlib, num_jitters=3)
    resA_thr = [(name if sim>=dlib_sim_thr else "Unknown", sim, bbox) for (name, sim, bbox) in resA]

    # InsightFace
    ins = InsightFaceWrapper(ctx_id=insight_ctx_id, det_size=insight_det_size)
    gal_ins = build_gallery_insightface(known_dir, ins)
    imgB, resB = recognize_with_insightface(image_path, gal_ins, ins)
    resB_thr = [(name if sim>=insight_sim_thr else "Unknown", sim, bbox) for (name, sim, bbox) in resB]

    print("=== dlib(face_recognition) results ===")
    for name, sim, _ in resA_thr:
        print(f"  {name:>16s} : {sim*100:6.2f}%")
    print("=== InsightFace(ArcFace) results ===")
    for name, sim, _ in resB_thr:
        print(f"  {name:>16s} : {sim*100:6.2f}%")

    draw_boxes_rgb(imgA, resA_thr, title="dlib (face_recognition) - cosine similarity")
    draw_boxes_rgb(imgB, resB_thr, title="InsightFace (RetinaFace + ArcFace) - cosine similarity")

# =======================
# 実行例（パスをあなたの環境に合わせて変更）
# =======================
# 構成例:
# known_faces/
#   alice/ a1.jpg a2.jpg
#   bob/   b1.png b2.png
# test_image.jpg

# compare_both(
#     known_dir="known_faces",
#     image_path="test_image.jpg",
#     dlib_sim_thr=0.70,            # face_recognition 側の閾値（0.65〜0.8で要検証）
#     insight_sim_thr=0.70,         # InsightFace 側の閾値（0.65〜0.8で要検証）
#     insight_ctx_id=0,             # GPU=0 / CPU=-1
#     insight_det_size=(640,640)    # 顔が小さいなら (960,960) に上げる
# )
```

```python
# =======================
# Jupyter 1セル実行用：両方式（dlib / InsightFace）比較 & 自動アップスケール & キャッシュ & 一括処理
# =======================
from pathlib import Path
import os, hashlib, time
import numpy as np
import cv2
import matplotlib.pyplot as plt
import face_recognition

# --------- 共通ユーティリティ ----------
VALID_EXTS = {".jpg", ".jpeg", ".png"}

def list_images(pathlike):
    p = Path(pathlike)
    if p.is_dir():
        files = []
        for ext in VALID_EXTS:
            files += list(p.glob(f"*{ext}"))
        return sorted(files)
    elif p.is_file() and p.suffix.lower() in VALID_EXTS:
        return [p]
    else:
        raise FileNotFoundError(f"画像が見つかりません: {p}")

def ensure_rgb_uint8(img):
    arr = np.asarray(img)
    if arr.ndim == 2:
        arr = np.stack([arr]*3, axis=-1)
    elif arr.ndim == 3 and arr.shape[2] == 4:
        arr = arr[:, :, :3]
    if arr.dtype != np.uint8:
        arr = arr.astype(np.uint8, copy=False)
    if not arr.flags["C_CONTIGUOUS"]:
        arr = np.ascontiguousarray(arr)
    return arr

def draw_results_on(img_rgb, results, color=(0,255,0)):
    canvas = img_rgb.copy()
    for label, score, (top, right, bottom, left) in results:
        cv2.rectangle(canvas, (left, top), (right, bottom), color, 2)
        txt = f"{label} ({score*100:.1f}%)"
        cv2.putText(canvas, txt, (left, max(0, top-8)), cv2.FONT_HERSHEY_SIMPLEX, 0.6, color, 2)
    return canvas

def show_side_by_side(img_left_rgb, res_left, img_right_rgb, res_right, title_left="dlib", title_right="ArcFace"):
    fig, axes = plt.subplots(1, 2, figsize=(12, 6))
    axes[0].imshow(draw_results_on(img_left_rgb, res_left, (0,255,0)))
    axes[0].set_title(title_left); axes[0].axis("off")
    axes[1].imshow(draw_results_on(img_right_rgb, res_right, (0,255,0)))
    axes[1].set_title(title_right); axes[1].axis("off")
    plt.tight_layout(); plt.show()

def l2_normalize(v, eps=1e-12):
    v = np.asarray(v)
    n = np.linalg.norm(v, axis=-1, keepdims=True)
    return v / np.clip(n, eps, None)

def cosine_sim(a, b):
    a = l2_normalize(a); b = l2_normalize(b)
    return np.dot(a, b.T)

# --------- known_faces 収集（隠しフォルダ除外） & ディレクトリ署名 ----------
def collect_known_images(known_dir: Path):
    if not known_dir.exists():
        raise FileNotFoundError(f"{known_dir} が存在しません")
    people = {}
    subdirs = [p for p in known_dir.iterdir() if p.is_dir() and not p.name.startswith(".")]
    if subdirs:
        for person_dir in sorted(subdirs):
            imgs = []
            for ext in VALID_EXTS:
                imgs += list(person_dir.glob(f"*{ext}"))
            imgs = [p for p in imgs if not p.name.startswith(".")]
            if imgs:
                people[person_dir.name] = imgs
    else:
        imgs = [p for p in known_dir.iterdir()
                if p.is_file() and p.suffix.lower() in VALID_EXTS and not p.name.startswith(".")]
        for p in imgs:
            people.setdefault(p.stem, []).append(p)
    return people

def dir_signature(known_dir: Path):
    """パスとmtimeから簡易署名を生成（内容変化検知）"""
    h = hashlib.sha1()
    for root, dirs, files in os.walk(known_dir):
        # 隠しは無視
        dirs[:] = [d for d in dirs if not d.startswith(".")]
        for f in sorted(files):
            if f.startswith("."): continue
            p = Path(root) / f
            if p.suffix.lower() not in VALID_EXTS: continue
            stat = p.stat()
            h.update(str(p).encode()); h.update(str(int(stat.st_mtime)).encode())
    return h.hexdigest()

# ==========================================================
#  A) dlib / face_recognition パイプライン（安全・TTA・自動アップスケール）
# ==========================================================
def safe_face_locations(img_rgb):
    locs = face_recognition.face_locations(img_rgb)
    return [(int(t), int(r), int(b), int(l)) for (t,r,b,l) in locs]

def encode_dlib_rgb_safe(img_rgb, face_locations=None, num_jitters=3, do_flip_tta=True):
    img_rgb = ensure_rgb_uint8(img_rgb)
    if face_locations is None:
        face_locations = safe_face_locations(img_rgb)
    else:
        face_locations = [(int(t), int(r), int(b), int(l)) for (t,r,b,l) in face_locations]
    encs = face_recognition.face_encodings(img_rgb, face_locations, num_jitters=num_jitters)
    encs = np.asarray(encs, dtype=np.float32)
    if do_flip_tta and len(encs):
        img_flip = img_rgb[:, ::-1, :].copy()
        w = img_rgb.shape[1]
        locs_flip = [(t, w - l, b, w - r) for (t,r,b,l) in face_locations]
        locs_flip = [(int(t), int(r), int(b), int(l)) for (t,r,b,l) in locs_flip]
        encs_flip = face_recognition.face_encodings(img_flip, locs_flip, num_jitters=1)
        if len(encs_flip) == len(encs):
            encs = (encs + np.asarray(encs_flip, dtype=np.float32)) / 2.0
    encs = l2_normalize(encs)
    return encs, face_locations

def auto_scale_dlib(img_rgb, target_min_face_px=80, scales=(1.0, 1.5, 2.0), num_jitters=3):
    """小顔対策：スケールを上げて検出→元座標へ戻す"""
    H, W = img_rgb.shape[:2]
    best = ([], [])
    for s in scales:
        if s == 1.0:
            img_s = img_rgb
        else:
            img_s = cv2.resize(img_rgb, (int(W*s), int(H*s)), interpolation=cv2.INTER_LINEAR)
        encs, locs = encode_dlib_rgb_safe(img_s, num_jitters=num_jitters, do_flip_tta=True)
        # 顔サイズ判定
        if len(locs):
            widths = [r - l for (t,r,b,l) in locs]
            if len(widths) and max(widths) >= target_min_face_px:
                # 元座標へ戻す
                if s != 1.0:
                    locs = [(int(t/s), int(r/s), int(b/s), int(l/s)) for (t,r,b,l) in locs]
                return encs, locs
            # 一応結果を保持（最後に返す）
            best = (encs, [(int(t/s), int(r/s), int(b/s), int(l/s)) if s!=1.0 else (t,r,b,l) for (t,r,b,l) in locs])
    return best  # 見つかった中で最大を返す（ない場合は空）

# テンプレキャッシュ
_DLIB_CACHE = {"sig": None, "templates": None, "dir": None}

def build_gallery_face_recognition_cached(known_dir: Path, num_jitters=1, force_rebuild=False):
    sig = dir_signature(known_dir)
    if (not force_rebuild) and _DLIB_CACHE["sig"] == sig and _DLIB_CACHE["dir"] == str(known_dir):
        return _DLIB_CACHE["templates"]
    people = collect_known_images(known_dir)
    templates = {}
    for name, paths in people.items():
        vecs = []
        for path in paths:
            img = face_recognition.load_image_file(str(path))
            img = ensure_rgb_uint8(img)
            encs, locs = auto_scale_dlib(img, target_min_face_px=80, scales=(1.0,1.5,2.0), num_jitters=1)
            if len(encs):
                vecs.append(np.mean(encs, axis=0))
        if vecs:
            v = np.mean(np.vstack(vecs), axis=0)
            templates[name] = l2_normalize(v)
    _DLIB_CACHE.update({"sig": sig, "templates": templates, "dir": str(known_dir)})
    print(f"[dlib] templates built: {list(templates.keys())}")
    return templates

def recognize_with_face_recognition_auto(img_rgb, templates: dict, dlib_sim_thr=0.70):
    encs, face_locations = auto_scale_dlib(img_rgb, target_min_face_px=80, scales=(1.0,1.5,2.0), num_jitters=3)
    names = list(templates.keys())
    gal = np.vstack([templates[n] for n in names]) if names else None
    results = []
    for q, (t,r,b,l) in zip(encs, face_locations):
        label, sim_best = "Unknown", 0.0
        if gal is not None and len(gal) > 0:
            sims = cosine_sim(q[None,:], gal)[0]
            j = int(np.argmax(sims)); sim_best = float(sims[j])
            if sim_best >= dlib_sim_thr:
                label = names[j]
        results.append((label, sim_best, (t,r,b,l)))
    return results

# ==========================================================
#  B) InsightFace（ArcFace + RetinaFace） 自動アップスケール & キャッシュ
# ==========================================================
try:
    from insightface.app import FaceAnalysis
except Exception as e:
    raise RuntimeError("InsightFaceのインポートに失敗。`pip install insightface onnxruntime-gpu`（CPUは onnxruntime）を実行してください。\n" + repr(e))

class InsightFaceWrapper:
    def __init__(self, ctx_id=0, det_size=(640,640), model_pack="buffalo_l"):
        self.app = FaceAnalysis(name=model_pack)
        self.app.prepare(ctx_id=ctx_id, det_size=det_size)

    def embed_bgr(self, bgr, do_flip_tta=True):
        faces = self.app.get(bgr)
        embs, boxes = [], []
        for f in faces:
            e = f.normed_embedding.astype(np.float32)  # L2済み
            if do_flip_tta:
                bgrf = cv2.flip(bgr, 1).copy()
                faces_f = self.app.get(bgrf)
                if faces_f:
                    e = (e + faces_f[0].normed_embedding.astype(np.float32)) / 2.0
                    e = e / (np.linalg.norm(e) + 1e-12)
            embs.append(e)
            boxes.append(f.bbox.astype(int))  # (l,t,r,b)
        if not embs:
            return np.zeros((0,512), np.float32), []
        return np.stack(embs), boxes

def auto_scale_insight(bgr, ins: InsightFaceWrapper, target_min_face_px=80, scales=(1.0,1.5,2.0)):
    H, W = bgr.shape[:2]
    best = (np.zeros((0,512),np.float32), [])
    for s in scales:
        if s == 1.0:
            img_s = bgr
        else:
            img_s = cv2.resize(bgr, (int(W*s), int(H*s)), interpolation=cv2.INTER_LINEAR)
        embs, boxes = ins.embed_bgr(img_s, do_flip_tta=True)
        if len(boxes):
            widths = [int(r - l) for (l,t,r,b) in boxes]
            if max(widths) >= target_min_face_px:
                if s != 1.0:
                    boxes = [(int(l/s), int(t/s), int(r/s), int(b/s)) for (l,t,r,b) in boxes]
                return embs, boxes
            best = (embs, [(int(l/s), int(t/s), int(r/s), int(b/s)) if s!=1.0 else (l,t,r,b) for (l,t,r,b) in boxes])
    return best

# テンプレキャッシュ
_INS_CACHE = {"sig": None, "templates": None, "dir": None, "ins": None, "ctx_id": None, "det_size": None}

def build_gallery_insightface_cached(known_dir: Path, ins: InsightFaceWrapper, force_rebuild=False):
    sig = dir_signature(known_dir)
    if (not force_rebuild) and _INS_CACHE["sig"] == sig and _INS_CACHE["dir"] == str(known_dir):
        return _INS_CACHE["templates"]
    people = collect_known_images(known_dir)
    templates = {}
    for name, paths in people.items():
        vecs = []
        for path in paths:
            bgr = cv2.imread(str(path))
            if bgr is None: continue
            embs, _ = auto_scale_insight(bgr, ins, target_min_face_px=80, scales=(1.0,1.5,2.0))
            if len(embs):
                vecs.append(embs[0])
        if vecs:
            v = np.mean(np.vstack(vecs), axis=0)
            templates[name] = l2_normalize(v)
    _INS_CACHE.update({"sig": sig, "templates": templates, "dir": str(known_dir)})
    print(f"[insightface] templates built: {list(templates.keys())}")
    return templates

def recognize_with_insightface_auto(bgr, templates: dict, ins: InsightFaceWrapper, insight_sim_thr=0.70):
    embs, boxes = auto_scale_insight(bgr, ins, target_min_face_px=80, scales=(1.0,1.5,2.0))
    names = list(templates.keys())
    gal = np.vstack([templates[n] for n in names]) if names else None
    results = []
    for e, (l,t,r,b) in zip(embs, boxes):
        label, sim_best = "Unknown", 0.0
        if gal is not None and len(gal) > 0:
            sims = np.dot(l2_normalize(e[None,:]), gal.T)[0]
            j = int(np.argmax(sims)); sim_best = float(sims[j])
            if sim_best >= insight_sim_thr:
                label = names[j]
        # to (top,right,bottom,left)
        results.append((label, sim_best, (int(t), int(r), int(b), int(l))))
    img_rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
    return img_rgb, results

# --------- メイン：一括比較 ----------
def compare_inputs(
    known_dir="known_faces",
    input_path="test_image.jpg",        # ファイル or ディレクトリ
    dlib_sim_thr=0.70,
    insight_sim_thr=0.70,
    insight_ctx_id=0,                   # GPU:0 / CPU:-1
    insight_det_size=(640,640)          # 小顔多いなら (960,960) など
):
    known_dir = Path(known_dir)
    files = list_images(input_path)
    if len(files) == 0:
        print("処理対象がありません"); return

    # キャッシュを使ってテンプレ構築
    gal_dlib = build_gallery_face_recognition_cached(known_dir, num_jitters=1, force_rebuild=False)
    ins = _INS_CACHE.get("ins")
    if (ins is None) or (_INS_CACHE.get("ctx_id") != insight_ctx_id) or (_INS_CACHE.get("det_size") != insight_det_size):
        ins = InsightFaceWrapper(ctx_id=insight_ctx_id, det_size=insight_det_size)
        _INS_CACHE["ins"] = ins; _INS_CACHE["ctx_id"] = insight_ctx_id; _INS_CACHE["det_size"] = insight_det_size
    gal_ins = build_gallery_insightface_cached(known_dir, ins, force_rebuild=False)

    for fp in files:
        print(f"\n=== {fp.name} ===")
        # dlib
        print(f"dlib")
        img = face_recognition.load_image_file(str(fp))
        img = ensure_rgb_uint8(img)
        res_dlib = recognize_with_face_recognition_auto(img, gal_dlib, dlib_sim_thr=dlib_sim_thr)

        # InsightFace
        print(f"insightface")
        bgr = cv2.imread(str(fp))
        if bgr is None:
            print("読み込み失敗（スキップ）"); continue
        img_ins_rgb, res_ins = recognize_with_insightface_auto(bgr, gal_ins, ins, insight_sim_thr=insight_sim_thr)

        # テキスト出力
        print("[dlib] results:")
        for name, sim, _ in res_dlib:
            print(f"  {name:>16s} : {sim*100:6.2f}%")
        print("[ArcFace] results:")
        for name, sim, _ in res_ins:
            print(f"  {name:>16s} : {sim*100:6.2f}%")

        # 可視化：左 dlib / 右 ArcFace
        show_side_by_side(img, res_dlib, img_ins_rgb, res_ins,
                          title_left="dlib (face_recognition)", title_right="InsightFace (ArcFace)")

# =======================
# 使い方サンプル
# =======================
# known_faces/
#   alice/ a1.jpg a2.jpg
#   bob/   b1.png b2.png
# compare_inputs(
#     known_dir="known_faces",
#     input_path="test_images",  # ディレクトリ or 単一ファイル
#     dlib_sim_thr=0.70,
#     insight_sim_thr=0.72,
#     insight_ctx_id=0,          # GPU:0 / CPU:-1
#     insight_det_size=(640,640) # 顔が小さいなら (960,960)
# )
```

```python
# ================================================
# FACE01 / face_recognition 切替・高速版・静音化・キャッシュ付 完全セル
# ================================================
from pathlib import Path
import os, io, hashlib, inspect, tempfile
from contextlib import redirect_stdout, redirect_stderr, contextmanager
import numpy as np
import cv2
import matplotlib.pyplot as plt
import face_recognition as fr
from face01lib.api import Dlib_api
from face01lib.api import Dlib_api

# ------- 可能な限りライブラリ出力を抑制 -------
os.environ.setdefault("OPENCV_LOG_LEVEL", "ERROR")
os.environ.setdefault("KMP_WARNINGS", "0")

@contextmanager
def _silence():
    """stdout/stderr を無音化するコンテキスト（FACE01での冗長出力抑制に使用）"""
    sio_out, sio_err = io.StringIO(), io.StringIO()
    with redirect_stdout(sio_out), redirect_stderr(sio_err):
        yield

# ------- 共通ユーティリティ -------
VALID_EXTS = {".jpg", ".jpeg", ".png"}
CACHE_DIR = Path("./.cache"); CACHE_DIR.mkdir(exist_ok=True, parents=True)

def ensure_rgb_uint8(img):
    arr = np.asarray(img)
    if arr.ndim == 2:
        arr = np.stack([arr]*3, axis=-1)
    elif arr.ndim == 3 and arr.shape[2] == 4:
        arr = arr[:, :, :3]
    if arr.dtype != np.uint8:
        arr = arr.astype(np.uint8, copy=False)
    if not arr.flags["C_CONTIGUOUS"]:
        arr = np.ascontiguousarray(arr)
    return arr

def _to_rgb(img_or_path):
    if isinstance(img_or_path, (str, Path)):
        return fr.load_image_file(str(img_or_path))  # RGB ndarray
    return ensure_rgb_uint8(img_or_path)

def l2_normalize(v, eps=1e-12):
    v = np.asarray(v, dtype=np.float32)
    n = np.linalg.norm(v, axis=-1, keepdims=True)
    return v / np.clip(n, eps, None)

def cosine_sim(a, b):
    a = l2_normalize(a); b = l2_normalize(b)
    return np.dot(a, b.T)

def _dir_signature(known_dir: Path):
    """known_dir の内容（相対パス・mtime・サイズ）から署名を作成（キャッシュキー）"""
    h = hashlib.sha1()
    for root, dirs, files in os.walk(known_dir):
        dirs[:] = [d for d in dirs if not d.startswith(".")]
        for f in sorted(files):
            if f.startswith("."): continue
            p = Path(root) / f
            if p.suffix.lower() not in VALID_EXTS: continue
            st = p.stat()
            h.update(str(p.relative_to(known_dir)).encode())
            h.update(str(int(st.st_mtime)).encode())
            h.update(str(int(st.st_size)).encode())
    return h.hexdigest()

def _collect_known_images(known_dir: Path, per_person_limit=3):
    """ known_dir を {name: [paths...]} にまとめる（1人あたり枚数制限）"""
    people = {}
    subdirs = [p for p in known_dir.iterdir() if p.is_dir() and not p.name.startswith(".")]
    if subdirs:
        for person_dir in sorted(subdirs):
            imgs = []
            for ext in VALID_EXTS:
                imgs += sorted(person_dir.glob(f"*{ext}"))
            imgs = [p for p in imgs if not p.name.startswith(".")][:per_person_limit]
            if imgs:
                people[person_dir.name] = imgs
    else:
        imgs = [p for p in known_dir.iterdir()
                if p.is_file() and p.suffix.lower() in VALID_EXTS and not p.name.startswith(".")]
        for p in sorted(imgs)[:per_person_limit * max(1, len(imgs))]:
            people.setdefault(p.stem, []).append(p)
    return people

def _build_person_thresholds_robust(templates: dict, percentile=0.98, floor=0.58, cap=0.84):
    """他人類似度の上位パーセンタイルをしきい値に（過大化防止でクリップ）"""
    names = list(templates.keys())
    if not names: return {}
    G = np.vstack([templates[n] for n in names]).astype(np.float32)
    thr = {}
    for i, name in enumerate(names):
        sims = (G[i:i+1] @ G.T).ravel()
        sims = np.delete(sims, i)
        if sims.size == 0:
            t = floor
        else:
            q = float(np.quantile(sims, percentile))
            q25, q75 = np.percentile(sims, [25, 75])
            iqr = q75 - q25
            iqr_cap = q75 + 1.5 * iqr
            t = min(q, iqr_cap)
        thr[name] = float(np.clip(t, floor, cap))
    return thr

# ------- face_recognition: 検出＋エンコード -------
def _safe_face_locations_fr(img_rgb):
    locs = fr.face_locations(img_rgb)
    return [(int(t), int(r), int(b), int(l)) for (t,r,b,l) in locs]

def _encode_rgb_fr(img_rgb, face_locations=None, num_jitters=1):
    img_rgb = ensure_rgb_uint8(img_rgb)
    if face_locations is None:
        face_locations = _safe_face_locations_fr(img_rgb)
    embs = fr.face_encodings(img_rgb, face_locations, num_jitters=num_jitters)
    embs = np.asarray(embs, dtype=np.float32)
    return l2_normalize(embs), face_locations

def _auto_scale_encode_fr(img_rgb, target_min_face_px=80, scales=(1.0,1.5,2.0), num_jitters=1):
    H, W = img_rgb.shape[:2]
    best = ([], [])
    for s in scales:
        img_s = img_rgb if s == 1.0 else cv2.resize(img_rgb, (int(W*s), int(H*s)), interpolation=cv2.INTER_LINEAR)
        embs, locs = _encode_rgb_fr(img_s, num_jitters=num_jitters)
        if len(locs):
            widths = [r - l for (t,r,b,l) in locs]
            if max(widths) >= target_min_face_px:
                if s != 1.0:
                    locs = [(int(t/s), int(r/s), int(b/s), int(l/s)) for (t,r,b,l) in locs]
                return embs, locs
            best = (embs, [(int(t/s), int(r/s), int(b/s), int(l/s)) if s!=1.0 else (t,r,b,l) for (t,r,b,l) in locs])
    return best

# ------- FACE01: Dlib_api を用いた高速エンコード（静音） -------
def _fix_bbox_order_for_face01(face_locations):
    fixed = []
    for (t, r, b, l) in face_locations:
        fixed.append((l, t, r, b))  # ← left, top, right, bottom に変換
    return fixed
def _face01_face_locations(img_rgb):
    """FACE01 の検出。失敗時は FR 検出へフォールバック"""
    try:
        api = Dlib_api()
        locs = api.face_locations(img_rgb)  # 想定: [(t,r,b,l), ...]
        return [(int(t), int(r), int(b), int(l)) for (t,r,b,l) in locs]
    except Exception:
        return _safe_face_locations_fr(img_rgb)

def _face01_face_encodings(img_rgb, face_locations, num_jitters=1):
    """FACE01 のエンコード。失敗時は FR へフォールバック"""
    try:
        # with _silence():
        api = Dlib_api()
        # embs = api.face_encodings(img_rgb, face_locations, num_jitters=num_jitters)
        locs_fixed = _fix_bbox_order_for_face01(face_locations)
        embs = api.face_encodings(img_rgb, locs_fixed, num_jitters=num_jitters)
        embs = np.asarray(embs, dtype=np.float32)
        return l2_normalize(embs)
    except Exception:
        embs = fr.face_encodings(img_rgb, face_locations, num_jitters=num_jitters)
        embs = np.asarray(embs, dtype=np.float32)
        return l2_normalize(embs)

# ------- ギャラリー構築（ディスクキャッシュ＋静音） -------
def build_gallery_cached(
    backend: str,                # "face01" or "fr"
    known_dir: str | Path,
    per_person_limit: int = 3,
    num_jitters: int = 1,
    force_rebuild: bool = False,
    thr_percentile: float = 0.98,
    thr_floor: float = 0.58,
    thr_cap: float = 0.84,
    quiet: bool = True
):
    """
    backend ごとに別キャッシュ:
      ./.cache/face01_gallery_<sig>.npz
      ./.cache/fr_gallery_<sig>.npz
    内容: 人名ごとの平均テンプレート（128/512 次元など）を npz に保存
    """
    known_dir = Path(known_dir)
    sig = _dir_signature(known_dir)
    suffix = "face01" if backend.lower().startswith("face01") else "fr"
    cache_path = CACHE_DIR / f"{suffix}_gallery_{sig}.npz"

    if cache_path.exists() and not force_rebuild:
        data = np.load(cache_path, allow_pickle=True)
        templates = {k: data[k].astype(np.float32) for k in data.files}
        person_thr = _build_person_thresholds_robust(templates, percentile=thr_percentile, floor=thr_floor, cap=thr_cap)
        return templates, person_thr

    # 再構築
    people = _collect_known_images(known_dir, per_person_limit=per_person_limit)
    templates = {}

    def _encode_face01(img):
        locs = _face01_face_locations(img)
        if not locs:
            embs, _ = _auto_scale_encode_fr(img, target_min_face_px=80, scales=(1.0,1.5,2.0), num_jitters=num_jitters)
            return embs
        return _face01_face_encodings(img, locs, num_jitters=num_jitters)

    def _encode_fr(img):
        locs = _safe_face_locations_fr(img)
        if not locs:
            embs, _ = _auto_scale_encode_fr(img, target_min_face_px=80, scales=(1.0,1.5,2.0), num_jitters=num_jitters)
            return embs
        embs, _ = _encode_rgb_fr(img, locs, num_jitters=num_jitters)
        return embs

    encode = _encode_face01 if suffix == "face01" else _encode_fr

    ctx = _silence() if quiet else None
    if ctx: ctx.__enter__()
    try:
        for name, paths in people.items():
            vecs = []
            for p in paths:
                img = _to_rgb(str(p))
                embs = encode(img)
                embs = np.asarray(embs, dtype=np.float32)
                if embs.size:
                    vecs.append(embs.mean(0))
            if vecs:
                v = np.mean(np.vstack(vecs), axis=0)
                v = v / (np.linalg.norm(v) + 1e-12)
                templates[name] = v.astype(np.float32)
    finally:
        if ctx: ctx.__exit__(None, None, None)

    if not templates:
        return {}, {}

    # しきい値計算 & 保存
    person_thr = _build_person_thresholds_robust(templates, percentile=thr_percentile, floor=thr_floor, cap=thr_cap)
    np.savez_compressed(cache_path, **templates)
    return templates, person_thr

# ------- 推論バックエンド -------
def _backend_face01_fast(img_rgb, known_dir=None,
                         base_min_sim=0.66, min_margin=0.01,
                         high_conf_override=0.97, topk_candidates=3):
    # 検出
    face_locations = _face01_face_locations(img_rgb)
    if not face_locations:
        _, face_locations = _auto_scale_encode_fr(img_rgb, target_min_face_px=80, scales=(1.0,1.5,2.0), num_jitters=1)
    print(f"fl : {face_locations}")
    if not face_locations:
        return []
    
    # エンコード
    embs = _face01_face_encodings(img_rgb, face_locations, num_jitters=1)
    print(f"embs : {embs}")
    if embs is None or len(embs) == 0:
        return []

    # ギャラリー
    templates, person_thr = ({}, {})
    if known_dir is not None:
        templates, person_thr = build_gallery_cached(
            backend="face01", known_dir=known_dir,
            per_person_limit=3, num_jitters=1, quiet=True
        )

    names = list(templates.keys())
    gal = np.vstack([templates[n] for n in names]) if names else None

    results = []
    for q, (t,r,b,l) in zip(embs, face_locations):
        label, sim_best, candidates = "Unknown", 0.0, []
        if gal is not None and len(gal)>0:
            sims = cosine_sim(q[None,:], gal)[0]
            order = np.argsort(-sims)
            topk = [(names[j], float(sims[j])) for j in order[:topk_candidates]]
            j = int(order[0]); top1 = float(sims[j]); top2 = float(sims[order[1]]) if len(order)>1 else -1.0
            if top1 >= high_conf_override:
                label, sim_best = names[j], top1
            else:
                ok = (top1 >= base_min_sim) and ((top1 - top2) >= min_margin) and \
                     (top1 >= person_thr.get(names[j], base_min_sim))
                label, sim_best = (names[j], top1) if ok else ("Unknown", top1)
            candidates = topk
        results.append({'box': (int(t), int(r), int(b), int(l)),
                        'name': label, 'similarity': float(sim_best), 'candidates': candidates})
    return results

def _backend_face_recognition(img_rgb, known_dir=None,
                              base_min_sim=0.66, min_margin=0.01,
                              high_conf_override=0.97, topk_candidates=3):
    embs, face_locations = _auto_scale_encode_fr(img_rgb, target_min_face_px=80,
                                                 scales=(1.0,1.5,2.0), num_jitters=1)
    if not face_locations:
        return []
    templates, person_thr = ({}, {})
    if known_dir is not None:
        templates, person_thr = build_gallery_cached(
            backend="fr", known_dir=known_dir,
            per_person_limit=2, num_jitters=1, quiet=True
        )
    names = list(templates.keys())
    gal = np.vstack([templates[n] for n in names]) if names else None

    results = []
    for q, (t,r,b,l) in zip(embs, face_locations):
        label, sim_best, candidates = "Unknown", 0.0, []
        if gal is not None and len(gal)>0:
            sims = cosine_sim(q[None,:], gal)[0]
            order = np.argsort(-sims)
            topk = [(names[j], float(sims[j])) for j in order[:topk_candidates]]
            j = int(order[0]); top1 = float(sims[j]); top2 = float(sims[order[1]]) if len(order)>1 else -1.0
            if top1 >= high_conf_override:
                label, sim_best = names[j], top1
            else:
                ok = (top1 >= base_min_sim) and ((top1 - top2) >= min_margin) and \
                     (top1 >= person_thr.get(names[j], base_min_sim))
                label, sim_best = (names[j], top1) if ok else ("Unknown", top1)
            candidates = topk
        results.append({'box': (int(t), int(r), int(b), int(l)),
                        'name': label, 'similarity': float(sim_best), 'candidates': candidates})
    return results

# ------- フロントAPI -------
def detect_faces(
    image,
    model: str = "face01",       # "face01"（推奨） or "face_recognition"
    known_dir: str | Path | None = None,
    min_sim: float = 0.66,       # 絶対値しきい値
    min_margin: float = 0.01,    # Top1-Top2 マージン
    high_conf_override: float = 0.97,
    topk_candidates: int = 3
):
    """
    Returns: list of dict:
      {'box':(top,right,bottom,left), 'name':str, 'similarity':float, 'candidates':[('name',score), ...]}
    """
    img_rgb = _to_rgb(image)
    m = model.lower()
    if m in ("face01", "face01_fast", "japanese_face"):
        return _backend_face01_fast(img_rgb, known_dir=known_dir,
                                    base_min_sim=min_sim, min_margin=min_margin,
                                    high_conf_override=high_conf_override,
                                    topk_candidates=topk_candidates)
    elif m in ("face_recognition", "fr", "dlib"):
        return _backend_face_recognition(img_rgb, known_dir=known_dir,
                                         base_min_sim=min_sim, min_margin=min_margin,
                                         high_conf_override=high_conf_override,
                                         topk_candidates=topk_candidates)
    else:
        raise ValueError(f"Unsupported model='{model}'. Use 'face01' or 'face_recognition'.")

# ------- 可視化 -------
def draw_results_on_image(rgb_img, results, box_color=(0,255,0), text_scale=0.7, text_thickness=2):
    vis = rgb_img.copy()
    for r in results:
        (t, rgt, btm, lft) = r['box']
        name = r.get('name', 'Unknown'); sim  = r.get('similarity', 0.0)
        cands = r.get('candidates', [])
        cv2.rectangle(vis, (lft, t), (rgt, btm), box_color, 2)
        label = f"{name} ({sim*100:.1f}%)"
        (tw, th), _ = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, text_scale, text_thickness)
        y1 = max(0, t - th - 6)
        cv2.rectangle(vis, (lft, y1), (lft + tw + 6, y1 + th + 6), (0, 0, 0), -1)
        cv2.putText(vis, label, (lft + 3, y1 + th + 3), cv2.FONT_HERSHEY_SIMPLEX, text_scale, (0,255,0), text_thickness)
        if name == "Unknown" and cands:
            offset = y1 + th + 8
            for (cand_name, score) in cands[:3]:
                txt = f"  → {cand_name} ({score*100:.1f}%)"
                cv2.putText(vis, txt, (lft + 3, offset), cv2.FONT_HERSHEY_SIMPLEX, text_scale*0.7, (0,200,255), 1)
                offset += int(th*0.9)
    return vis

def visualize_detection(image, res, title=None, dpi=120):
    rgb = _to_rgb(image)
    vis = draw_results_on_image(rgb, res)
    plt.figure(figsize=(8, 8), dpi=dpi)
    plt.imshow(vis); plt.axis("off")
    if title: plt.title(title)
    plt.show()
```

```python
# === Cell: build gallery ===
import os, glob
from collections import defaultdict
import numpy as np

from face01lib.api import Dlib_api
from face01lib.Calc import Cal

API = Dlib_api()
CAL = Cal()

KNOWN_DIR = "../nogi_member_face/"
VALID_EXTS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")
DEEP_MODEL = 1  # 0=dlib, 1=efficientnetv2_arcface, 2=mobilefacenet  ← ArcFaceを使用

def iter_known_images(root=KNOWN_DIR):
    for path in glob.glob(os.path.join(root, "**", "*"), recursive=True):
        if path.lower().endswith(VALID_EXTS) and os.path.isfile(path):
            label = os.path.basename(os.path.dirname(path))  # 親ディレクトリ名を人物名に
            yield label, path

def encode_image_file_safe(img_path, api, num_jitters=0):
    img = api.load_image_file(img_path)
    if img is None or img.ndim != 3:
        print(f"[WARN] failed to load image: {img_path}")
        return [], []
    locs = api.face_locations(img)
    if not locs:
        print(f"[WARN] no face found: {img_path}")
        return [], []
    try:
        # embs = api.face_encodings(img, locs, num_jitters=num_jitters)
        embs = api.face_encodings(deep_learning_model=1, resized_frame = img, face_location_list = locs)
        # print(f"embs : {embs}")
        embs = [np.asarray(e, dtype=np.float32) for e in embs]
        return locs, embs
    except Exception as e:
        print(f"[ERR] encoding failed: {e}")
        return [], []


# ギャラリー：label -> {"embs":[...], "paths":[...]}
gallery = defaultdict(lambda: {"embs": [], "paths": []})

skipped = []
for label, path in iter_known_images(KNOWN_DIR):
    locs, embs = encode_image_file_safe(path, API, num_jitters=0) 
    if not embs:
        skipped.append(path)
        continue
    # 1枚に複数顔があっても先頭のみ採用（必要なら全採用に変更可）
    gallery[label]["embs"].append(embs[0])
    gallery[label]["paths"].append(path)

# ラベルごとの重心ベクトル（平均）も用意：素早い比較用
label_centroids = {}
for label, data in gallery.items():
    arr = np.stack(data["embs"], axis=0)  # (N, D)
    label_centroids[label] = arr.mean(axis=0).astype(np.float32)

print(f"Loaded labels: {list(gallery.keys())}")
if skipped:
    print(f"[warn] No face detected in {len(skipped)} images (skipped).")
```

```python
# === Cell: inference ===
import numpy as np

# 類似度しきい値（%）。この値未満は "Unknown" にする
SIM_THRESH = 75.0  # 例: 75%（必要に応じて調整）

def distance_to_percent(dist: float, deep_model: int = DEEP_MODEL) -> float:
    """FACE01のCalで距離→類似度(%)に変換。"""
    return float(CAL.return_percentage(float(dist), deep_model))

def recognize_faces(query_image_path: str, topk: int = 3, use_centroid_first=True):
    """
    画像中の各顔に対して最尤ラベルを推定し、可視化用resultsを返す。
    results: [{"box":(t,r,b,l), "name":str, "similarity":float, "candidates":[(name,sim), ...]}, ...]
    """
    img = API.load_image_file(query_image_path)
    locs = API.face_locations(img)
    if not locs:
        return img, []

    embs = API.face_encodings(img, locs, num_jitters=0)
    embs = API.face_encodings(deep_learning_model=1, resized_frame = img, face_location_list = locs)
    
    embs = [np.asarray(e, dtype=np.float32) for e in embs]

    results = []
    labels = list(gallery.keys())
    if not labels:
        return img, results

    # まず重心で粗くスクリーニング → 必要なら全埋め込みで精密比較
    centroid_mat = None
    if use_centroid_first:
        centroid_mat = np.stack([label_centroids[lbl] for lbl in labels], axis=0)  # (L, D)

    for box, q in zip(locs, embs):
        # 候補スコア計算
        cand_scores = []
        print(box, q)
        if use_centroid_first and centroid_mat is not None:
            dists = np.linalg.norm(centroid_mat - q[None, :], axis=1)
            dists = np.asarray(dists).reshape(-1)   # flatten保証
            order = np.argsort(dists)[:max(5, topk+2)]
            narrowed = [labels[int(i)] for i in order.tolist()]


        else:
            narrowed = labels

        for lbl in narrowed:
            # ラベル下の全埋め込みと最小距離を取る（強い最近傍ルール）
            known = np.stack(gallery[lbl]["embs"], axis=0)  # (N,D)
            d = np.linalg.norm(known - q[None, :], axis=1).min()
            sim = distance_to_percent(d, DEEP_MODEL)
            cand_scores.append((lbl, sim))

        # 類似度の高い順に並べる
        cand_scores.sort(key=lambda x: x[1], reverse=True)
        best_name, best_sim = cand_scores[0]

        out_name = best_name if best_sim >= SIM_THRESH else "Unknown"
        results.append({
            "box": box,  # (top, right, bottom, left)
            "name": out_name,
            "similarity": best_sim/100.0,          # あなたの可視化が 0-1 を想定しているので%
            "candidates": [(n, s/100.0) for (n, s) in cand_scores[:topk]],
        })
    return img, results
```

```python
from face01lib.api import Dlib_api
from face01lib.Calc import Cal
import numpy as np, os, glob

# モデル番号: 0 = dlib, 1 = JAPANESE_FACE_V1.onnx
DEEP_MODEL = 1

# 入力・閾値
KNOWN_DIR   = "../nogi_member_face/"
VALID_EXTS  = (".jpg",".jpeg",".png",".bmp",".webp")
SIM_THRESH  = 70.0  # Cal.return_percentage の結果(%)で Unknown 判定の閾値

API = Dlib_api()
CAL = Cal()

```

```python
# === Cell 2: build gallery (force (512,) vectors) ===
def build_gallery(root=KNOWN_DIR):
    known_encs, known_names = [], []
    if not os.path.isdir(root):
        raise FileNotFoundError(f"KNOWN_DIR not found: {root}")

    for label in sorted(os.listdir(root)):
        d = os.path.join(root, label)
        if not os.path.isdir(d): 
            continue
        for p in glob.glob(os.path.join(d, "*")):
            if not p.lower().endswith(VALID_EXTS):
                continue
            img  = API.load_image_file(p)
            locs = API.face_locations(img)  # [(t,r,b,l), ...]
            if not locs:
                continue
            # 重要: 第1引数は deep_learning_model
            embs = API.face_encodings(DEEP_MODEL, img, locs, num_jitters=0)
            if not embs:
                continue
            v = np.asarray(embs[0], dtype=np.float64).ravel()  # (512,)
            known_encs.append(v)
            known_names.append(label)

    return [np.asarray(e, np.float64).ravel() for e in known_encs], known_names

KNOWN_ENCS, KNOWN_NAMES = build_gallery()
print(f"gallery samples = {len(KNOWN_NAMES)} / labels = {sorted(set(KNOWN_NAMES))}")
```

```python
# === Cell 3: recognize (L2 distance -> percentage) ===
def recognize_image(image_path, topk=3):
    """
    Returns:
        rgb_img, results
    results[i] = {
        "box": (t, r, b, l),
        "name": "b"/"d"/"Unknown",
        "similarity": 0..1,                 # 0-1 (可視化用)
        "candidates": [(name, 0..1), ...],  # 上位候補
    }
    """
    img  = API.load_image_file(image_path)
    locs = API.face_locations(img)
    if not locs or not KNOWN_ENCS:
        return img, []

    q_embs = API.face_encodings(DEEP_MODEL, img, locs, num_jitters=0)
    if not q_embs:
        return img, []

    # K: (N,512) を保証
    K = np.vstack([np.asarray(e, dtype=np.float64).ravel() for e in KNOWN_ENCS])  # (N,512)
    N, D = K.shape

    results = []
    for box, q in zip(locs, q_embs):
        q = np.asarray(q, dtype=np.float64).ravel()  # (512,)
        if q.ndim != 1:
            q = q.reshape(-1)
        if q.shape[0] != D:
            raise ValueError(f"embedding dim mismatch: K.shape={K.shape}, q.shape={q.shape}")

        # --- L2距離 (N,) ---
        dists = np.linalg.norm(K - q[None, :], axis=1)

        # ラベルごとに最小距離を採用 → 距離→％に変換
        best_by_label = {}  # name -> (best_dist, best_pct)
        for d, name in zip(dists, KNOWN_NAMES):
            d = float(d)
            if name not in best_by_label or d < best_by_label[name][0]:
                pct = float(CAL.return_percentage(d, DEEP_MODEL))  # 0..100
                best_by_label[name] = (d, pct)

        # 上位候補（％降順）
        cand = sorted(((n, pct) for n, (_, pct) in best_by_label.items()),
                      key=lambda x: x[1], reverse=True)[:max(1, topk)]
        best_name, best_pct = cand[0]
        out_name = best_name if best_pct >= SIM_THRESH else "Unknown"

        results.append({
            "box": box,
            "name": out_name,
            "similarity": best_pct / 100.0,
            "candidates": [(n, p/100.0) for (n, p) in cand],
        })

    return img, results
```
